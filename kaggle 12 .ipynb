# Obesity Risk Prediction for Cardiovascular Disease

## Introduction
This project aims to predict obesity risk in individuals, a key factor related to cardiovascular disease, using various health-related features. The dataset includes demographic and medical information that will be used to develop a predictive model. The data can be downloaded from the [Kaggle competition page](https://www.kaggle.com/competitions/playground-series-s4e2/data).


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split , GridSearchCV
from sklearn.metrics import classification_report , accuracy_score
from sklearn.preprocessing import LabelEncoder , StandardScaler
from sklearn.ensemble import GradientBoostingClassifier , RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier
```

```python
train = pd.read_csv("train.csv")
test = pd.read_csv("test.csv")
```

```python
le = LabelEncoder()
```

```python
obj = train.select_dtypes(include='object')
non_obj = train.select_dtypes(exclude='object')
```

```python
for i in range(0, obj.shape[1]):
    obj.iloc[:, i] = le.fit_transform(obj.iloc[:, i])
```

```python
obj = obj.astype("int")
```

```python
train = pd.concat([obj, non_obj], axis=1)
```

```python
obj = test.select_dtypes(include='object')
non_obj = test.select_dtypes(exclude='object')
for i in range(0, obj.shape[1]):
    obj.iloc[:, i] = le.fit_transform(obj.iloc[:, i])
obj = obj.astype("int")
test = pd.concat([obj, non_obj], axis=1)
```

```python
scaler = StandardScaler()
```

```python
scal_1 = train[['Age']]
scal_2 = train[['Weight']]
train['Age'] = scaler.fit_transform(scal_1)
train['Weight'] = scaler.fit_transform(scal_2)
```

```python
scal_1 = test[['Age']]
scal_2 = test[['Weight']]
test['Age'] = scaler.fit_transform(scal_1)
test['Weight'] = scaler.fit_transform(scal_2)
```

```python
x = train.drop(['NObeyesdad','id'],axis=1)
y = train['NObeyesdad']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
lr= LogisticRegression()
gb= GradientBoostingClassifier()
svc= SVC()
gas= GaussianNB()
knn= KNeighborsClassifier()
xb= XGBClassifier()
```

```python
Algorthim = ['LogisticRegression()',' GradientBoostingClassifier()', 'SVC()', 'GaussianNB()', 'KNeighborsClassifier()', 'XGBClassifier()']
Result_1=[]
```

```python
def Model(Model):
    Model.fit(x_train,y_train)
    pre = Model.predict(x_test)
    accuracy = accuracy_score(pre, y_test)
    print("Model is: ",Model)
    print("Accuracy is: ",round((accuracy*100),2),"%")
    print(classification_report(pre, y_test))
    Result_1.append(accuracy)
```

```python
Model(lr)
```

```python
Model(gb)
Model(svc)
Model(gas)
Model(knn)
Model(xb)
```

```python
Final = pd.DataFrame({"Algorithm": Algorthim,
                     "Result: ": Result_1})
print(Final)
print("----------------------------")
print("The Best Algo is: ",Final.max())
```

```python
test_x = test.drop('id', axis=1)
```

```python
pred_x = xb.predict(test_x)
```

```python
submission = pd.DataFrame({"id":test['id'],
                           "NObeyesdad":pred_x})
```

```python
mapping_dict = {
    4: 'Obesity_Type_III',
    3: 'Obesity_Type_II',
    1: 'Normal_Weight',
    2: 'Obesity_Type_I',
    0: 'Insufficient_Weight',
    6: 'Overweight_Level_II',
    5: 'Overweight_Level_I'
}
submission['NObeyesdad'] = submission['NObeyesdad'].map(mapping_dict)
```

```python
submission.to_csv("submission.csv",index=False)
```

## Conclusion

In this project, multiple machine learning models were evaluated to predict obesity risk, which is closely related to cardiovascular disease. Various models were tested, including Logistic Regression, Gradient Boosting Classifier, Support Vector Classifier (SVC), Gaussian Naive Bayes, K-Nearest Neighbors (KNN), and XGBoost Classifier. 

- The **XGBoost Classifier** delivered the best performance, achieving an accuracy of **90.03%** on the validation set, and was selected as the final model for the competition. This tree-based model performed exceptionally well in capturing the underlying patterns of the dataset.
- The **Gradient Boosting Classifier** also performed strongly with an accuracy of **90.27%**, indicating that boosting algorithms are highly effective for this task.
- Other models like **Logistic Regression** and **SVC** provided good results with accuracies of **80.42%** and **85.24%**, respectively, but did not perform as well as the boosting models.
- The **Gaussian Naive Bayes** model struggled with an accuracy of **66.31%**, and the **KNN** model, while better, had an accuracy of **77.05%**.

The final Kaggle result of **0.88908** suggests that the model generalizes well to unseen data, though there is still room for improvement. To further refine the model, hyperparameter tuning, feature engineering, and additional techniques such as ensemble methods could be explored. Overall, the XGBoost model demonstrated strong predictive power and is the best model for this task based on both the validation and competition results.
