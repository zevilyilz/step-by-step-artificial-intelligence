# Predicting Crab Age Based on Physical Attributes  

## Introduction  
In this project, the goal is to predict the age of crabs using their physical attributes through regression techniques. The project utilizes AutoML to automate model selection and tuning, aiming for the best predictive performance. The results will be evaluated using the provided test dataset and the predictions will be saved in a submission file for further use.  
[Download the dataset here.](https://www.kaggle.com/competitions/playground-series-s3e16/data)

```python
import h2o
import pandas as pd
from h2o.estimators import H2OGeneralizedLinearEstimator
from sklearn.model_selection import train_test_split

# Initialize H2O cluster
h2o.init()

# Load the datasets
train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')

# Convert the pandas DataFrames to H2OFrames
train_h2o = h2o.H2OFrame(train)
test_h2o = h2o.H2OFrame(test)

# Handle categorical columns by converting them to factors (H2O automatically handles categorical variables)
train_h2o['Sex'] = train_h2o['Sex'].asfactor()  # Convert 'Sex' column to a factor (example, add other categorical columns)
test_h2o['Sex'] = test_h2o['Sex'].asfactor()

# Separate the target variable 'Age' from features in train dataset
X_train = train_h2o.drop('Age')  # Drop 'Age' column (no need for 'columns=' argument)
y_train = train_h2o['Age']

# Initialize the H2O GLM (Generalized Linear Model) for regression
glm_model = H2OGeneralizedLinearEstimator(family="gaussian", lambda_=0.1)

# Train the model on the training data
glm_model.train(x=X_train.columns, y='Age', training_frame=train_h2o)

# Predict the target variable for the test data
predictions = glm_model.predict(test_h2o)

# Prepare the submission DataFrame
submission = pd.DataFrame({
    'id': test['id'],  # Assuming the test set has an 'id' column for reference
    'PredictedAge': predictions.as_data_frame()['predict']
})

# Save the predictions in a CSV file
submission.to_csv('submission.csv', index=False)

# Shutdown the H2O cluster after use
h2o.shutdown()
```
