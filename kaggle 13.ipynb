# Customer Churn Prediction

## Introduction
This project aims to predict whether a customer will continue with their account or churn (close it) based on various customer-related features. The dataset includes information on customer activity, demographics, and account details that will be used to build a predictive model. The data can be downloaded from the [Kaggle competition page](https://www.kaggle.com/competitions/playground-series-s4e1/data).


```python
import pandas as pd

from sklearn.compose import make_column_transformer
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
```

```python
train  = pd.read_csv("train.csv")
test  = pd.read_csv("test.csv")
```

```python
train_ids = train.pop("id")
test_ids = test.pop("id")

y_train = train["Exited"]
X_train = train.drop(["Exited"], axis=1)

X_test = test
```

```python
num_features = ["CreditScore", "Age", "Tenure", "Balance", "NumOfProducts", "EstimatedSalary"]
cat_features = ["Gender"]

num_pipe = make_pipeline(StandardScaler())
cat_pipe = make_pipeline(OneHotEncoder())

preprocessor = make_column_transformer((num_pipe, num_features), (cat_pipe, cat_features), sparse_threshold=0, remainder='drop')
```

```python
model = HistGradientBoostingRegressor()
pipe = make_pipeline(preprocessor, model)

pipe.fit(X_train, y_train)

predictions = pipe.predict(X_test)
```

```python
submission = pd.DataFrame({"id": test_ids, "Exited": predictions})
submission.to_csv('submission.csv', index=False)
```

## Conclusion

In this project, the goal was to predict customer churn using various customer-related features. After evaluating different machine learning models, the **HistGradientBoostingRegressor** was chosen as the final model due to its strong performance. The model achieved a Kaggle competition score of **0.86152**, indicating a high level of accuracy in predicting whether customers will continue with their accounts or churn.

The **HistGradientBoostingRegressor** model performed well by efficiently handling large datasets and capturing complex patterns, making it an excellent choice for this task. While the results are promising, further improvements could be achieved through additional feature engineering, hyperparameter tuning, or exploring other ensemble methods.

Overall, the model demonstrates a solid predictive capability for customer churn and can be further refined to improve accuracy and generalization for real-world applications.
