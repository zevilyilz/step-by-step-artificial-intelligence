# Predicting Mineral Hardness Using H2O AutoML

## Introduction  
In this project, we predict the Mohs hardness of minerals based on their physical and chemical properties using H2O AutoML. The dataset can be downloaded from the [Kaggle Playground Series Season 3, Episode 25](https://www.kaggle.com/competitions/playground-series-s3e25/data).


```python
pip install h2o
```

```python
import h2o
from h2o.automl import H2OAutoML
import pandas as pd

# Initialize H2O
h2o.init()

# Load datasets
train_data = pd.read_csv("train.csv")
test_data = pd.read_csv("test.csv")

# Convert pandas DataFrames to H2O Frames
train_h2o = h2o.H2OFrame(train_data)
test_h2o = h2o.H2OFrame(test_data)

# Set predictor and response variables
x = train_h2o.columns[1:-1]  # All columns except 'id' and 'Hardness'
y = "Hardness"

# Ensure the response column is treated as numeric
train_h2o[y] = train_h2o[y].asnumeric()

# Run H2O AutoML
aml = H2OAutoML(max_models=20, seed=42, verbosity="info")
aml.train(x=x, y=y, training_frame=train_h2o)

# Predict on the test data
predictions = aml.predict(test_h2o)

# Prepare the submission DataFrame
submission = test_data[["id"]].copy()
submission["Hardness"] = h2o.as_list(predictions)["predict"]

# Save submission file
submission.to_csv("submission.csv", index=False)

# Shutdown H2O
h2o.shutdown(prompt=False)
```

### Conclusion

The **H2O AutoML** approach for predicting Mohs hardness achieved a Kaggle score of **0.61588**, significantly improving model performance. The best model in this case was the **StackedEnsemble**, combining multiple models to deliver a lower RMSE and better overall predictions.
